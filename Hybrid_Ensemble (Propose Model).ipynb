```python
"""
Leakage-mitigated Hybrid Ensemble for Brain Tumor Classification
EfficientNet-B0 + ViT-Base with Cross-Modal Attention + Adaptive Fusion Gate + MC Dropout

Key fixes for Reviewer #1/#2:
1) Exact duplicate detection (md5)
2) Near-duplicate clustering (pHash) -> pseudo-subject groups
3) Group-wise train/val/test split (no group overlap)
4) Re-train + report stronger metrics + uncertainty evidence

Enhanced with:
- FIXED MC Dropout for proper uncertainty quantification
- Confusion Matrix visualization
- Loss & Accuracy curves
- ROC-AUC curves (macro/micro)
- Attention Map visualization
- Uncertainty statistics with proper variance
- Bootstrap Confidence Interval visualization
"""

import os, glob, random, hashlib
from collections import defaultdict

import numpy as np
from PIL import Image

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

from torchvision import transforms
import timm

from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import (
    accuracy_score, classification_report, confusion_matrix,
    precision_recall_fscore_support, roc_curve, auc, roc_auc_score
)

import matplotlib.pyplot as plt
import seaborn as sns

# ----------------------------
# Reproducibility
# ----------------------------
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# ----------------------------
# Config (ORIGINAL - UNCHANGED)
# ----------------------------
class Config:
    data_root = "/kaggle/input/brain-tumor-mri-dataset"  # dataset root
    img_size = 224
    batch_size = 16
    epochs = 30
    lr = 1e-4
    weight_decay = 1e-4

    num_classes = 4
    fusion_dim = 512

    mc_dropout_samples = 10  # ORIGINAL VALUE
    dropout_rate = 0.3  # ORIGINAL VALUE

    # pHash clustering
    phash_distance_threshold = 6

config = Config()

# ----------------------------
# Utils: file discovery
# ----------------------------
def list_images(root):
    exts = ("*.jpg", "*.jpeg", "*.png", "*.bmp")
    files = []
    for ext in exts:
        files.extend(glob.glob(os.path.join(root, "**", ext), recursive=True))
    return sorted(files)

def get_label_from_path(p):
    return os.path.basename(os.path.dirname(p))

# ----------------------------
# Step 1: Exact duplicate detection (md5)
# ----------------------------
def file_md5(path, chunk_size=1024 * 1024):
    m = hashlib.md5()
    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            m.update(chunk)
    return m.hexdigest()

# ----------------------------
# Step 2: Near-duplicate clustering (pHash)
# ----------------------------
def compute_phash(img_path):
    import imagehash
    img = Image.open(img_path).convert("RGB")
    return imagehash.phash(img)

def build_near_duplicate_clusters(all_images, threshold=6):
    import imagehash
    phashes = {}
    for p in all_images:
        phashes[p] = compute_phash(p)

    clusters = defaultdict(list)
    used = set()

    for i, a in enumerate(all_images):
        if a in used:
            continue
        used.add(a)
        clusters[a].append(a)
        ha = phashes[a]
        for b in all_images[i+1:]:
            if b in used:
                continue
            if ha - phashes[b] <= threshold:
                clusters[a].append(b)
                used.add(b)

    return clusters

# ----------------------------
# Dataset: file list
# ----------------------------
class FileListDataset(Dataset):
    def __init__(self, img_paths, class_to_idx, transform=None):
        self.img_paths = img_paths
        self.class_to_idx = class_to_idx
        self.transform = transform

    def __len__(self):
        return len(self.img_paths)

    def __getitem__(self, idx):
        p = self.img_paths[idx]
        y_name = get_label_from_path(p)
        y = self.class_to_idx[y_name]
        img = Image.open(p).convert("RGB")
        if self.transform:
            img = self.transform(img)
        return img, y

# ----------------------------
# Model components (ORIGINAL STRUCTURE - ONLY MC DROPOUT FIXED)
# ----------------------------
class CrossModalAttention(nn.Module):
    def __init__(self, cnn_dim, vit_dim, hidden_dim=512, num_heads=8):
        super().__init__()
        self.cnn_proj = nn.Linear(cnn_dim, hidden_dim)
        self.vit_proj = nn.Linear(vit_dim, hidden_dim)

        self.cross_attn_cnn_to_vit = nn.MultiheadAttention(
            hidden_dim, num_heads, dropout=0.1, batch_first=True
        )
        self.cross_attn_vit_to_cnn = nn.MultiheadAttention(
            hidden_dim, num_heads, dropout=0.1, batch_first=True
        )

        self.ln1 = nn.LayerNorm(hidden_dim)
        self.ln2 = nn.LayerNorm(hidden_dim)

    def forward(self, cnn_feat, vit_feat, return_attention=False):
        cnn_proj = self.cnn_proj(cnn_feat).unsqueeze(1)
        vit_proj = self.vit_proj(vit_feat).unsqueeze(1)

        cnn_enh, cnn_attn = self.cross_attn_cnn_to_vit(cnn_proj, vit_proj, vit_proj)
        cnn_enh = self.ln1(cnn_enh + cnn_proj).squeeze(1)

        vit_enh, vit_attn = self.cross_attn_vit_to_cnn(vit_proj, cnn_proj, cnn_proj)
        vit_enh = self.ln2(vit_enh + vit_proj).squeeze(1)

        if return_attention:
            return cnn_enh, vit_enh, cnn_attn, vit_attn
        return cnn_enh, vit_enh

class AdaptiveFusionGate(nn.Module):
    def __init__(self, feature_dim):
        super().__init__()
        self.gate = nn.Sequential(
            nn.Linear(feature_dim * 2, feature_dim),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(feature_dim, 2),
            nn.Softmax(dim=1),
        )

    def forward(self, cnn_feat, vit_feat):
        z = torch.cat([cnn_feat, vit_feat], dim=1)
        w = self.gate(z)
        cnn_w = w[:, 0:1]
        vit_w = w[:, 1:2]
        fused = cnn_w * cnn_feat + vit_w * vit_feat
        return fused, w

class UncertaintyHead(nn.Module):
    def __init__(self, input_dim, num_classes, dropout_rate=0.3):
        super().__init__()
        self.fc1 = nn.Linear(input_dim, 256)
        self.drop1 = nn.Dropout(dropout_rate)
        self.fc2 = nn.Linear(256, 128)
        self.drop2 = nn.Dropout(dropout_rate)
        self.fc3 = nn.Linear(128, num_classes)

    def forward(self, x, mc_dropout=False):
        x = F.relu(self.fc1(x))
        x = self.drop1(x) if (self.training or mc_dropout) else x
        x = F.relu(self.fc2(x))
        x = self.drop2(x) if (self.training or mc_dropout) else x
        return self.fc3(x)

    def predict_with_uncertainty(self, x, n_samples=10):
        """FIXED: MC Dropout for uncertainty estimation"""
        # Store original training state
        was_training = self.training
        
        # CRITICAL FIX: Enable training mode to activate dropout
        self.train()
        
        preds = []
        for _ in range(n_samples):
            with torch.no_grad():
                logits = self.forward(x, mc_dropout=True)
                preds.append(F.softmax(logits, dim=1))
        
        # Restore original state
        if not was_training:
            self.eval()
            
        preds = torch.stack(preds, dim=0)
        mean_prob = preds.mean(dim=0)
        uncertainty = preds.var(dim=0).mean(dim=1)
        return mean_prob, uncertainty

class HybridEnsemble(nn.Module):
    def __init__(self, num_classes=4, pretrained=True, fusion_dim=512, dropout_rate=0.3):
        super().__init__()

        self.efficientnet = timm.create_model(
            "efficientnet_b0", pretrained=pretrained, num_classes=0, global_pool=""
        )
        self.cnn_pool = nn.AdaptiveAvgPool2d(1)
        cnn_dim = self.efficientnet.num_features

        self.vit = timm.create_model(
            "vit_base_patch16_224", pretrained=pretrained, num_classes=0
        )
        vit_dim = self.vit.num_features

        self.cross_attention = CrossModalAttention(
            cnn_dim=cnn_dim, vit_dim=vit_dim, hidden_dim=fusion_dim
        )
        self.fusion_gate = AdaptiveFusionGate(fusion_dim)
        self.classifier = UncertaintyHead(fusion_dim, num_classes, dropout_rate=dropout_rate)

    def forward(self, x, return_attention=False):
        cnn_feat = self.efficientnet(x)
        cnn_feat = self.cnn_pool(cnn_feat).flatten(1)

        vit_feat = self.vit.forward_features(x)
        vit_feat = vit_feat[:, 0]

        if return_attention:
            cnn_enh, vit_enh, cnn_attn, vit_attn = self.cross_attention(
                cnn_feat, vit_feat, return_attention=True
            )
        else:
            cnn_enh, vit_enh = self.cross_attention(cnn_feat, vit_feat)
            cnn_attn, vit_attn = None, None

        fused, fusion_weights = self.fusion_gate(cnn_enh, vit_enh)
        logits = self.classifier(fused)

        if return_attention:
            return logits, cnn_attn, vit_attn, fusion_weights
        return logits

    def predict_with_uncertainty(self, x, mc_samples=10):
        """FIXED: Proper MC Dropout implementation"""
        self.eval()
        with torch.no_grad():
            cnn_feat = self.efficientnet(x)
            cnn_feat = self.cnn_pool(cnn_feat).flatten(1)
            vit_feat = self.vit.forward_features(x)
            vit_feat = vit_feat[:, 0]
            cnn_enh, vit_enh = self.cross_attention(cnn_feat, vit_feat)
            fused, _ = self.fusion_gate(cnn_enh, vit_enh)

        mean_prob, uncert = self.classifier.predict_with_uncertainty(fused, n_samples=mc_samples)
        return mean_prob, uncert

# ----------------------------
# Loss: Focal + label smoothing (ORIGINAL - UNCHANGED)
# ----------------------------
class FocalLossWithSmoothing(nn.Module):
    def __init__(self, alpha=1.0, gamma=2.0, smoothing=0.1):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma
        self.smoothing = smoothing

    def forward(self, logits, target):
        log_probs = F.log_softmax(logits, dim=-1)
        probs = torch.exp(log_probs)
        n_classes = logits.size(-1)

        with torch.no_grad():
            true_dist = torch.zeros_like(log_probs)
            true_dist.fill_(self.smoothing / (n_classes - 1))
            true_dist.scatter_(1, target.unsqueeze(1), 1.0 - self.smoothing)

        focal_w = (1.0 - probs) ** self.gamma
        loss = -self.alpha * focal_w * true_dist * log_probs
        return loss.sum(dim=1).mean()

# ----------------------------
# Transforms (ORIGINAL - UNCHANGED)
# ----------------------------
train_transform = transforms.Compose([
    transforms.RandomResizedCrop(config.img_size, scale=(0.8, 1.0)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.5),
    transforms.RandomRotation(20),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

eval_transform = transforms.Compose([
    transforms.Resize((config.img_size, config.img_size)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

# ----------------------------
# Build leakage-free splits (ORIGINAL - UNCHANGED)
# ----------------------------
print("\n[1] Listing images...")
all_images = list_images(config.data_root)
print("Total discovered images:", len(all_images))

labels_all = [get_label_from_path(p) for p in all_images]
unique_classes = sorted(list(set(labels_all)))
print("Detected classes:", unique_classes)

class_to_idx = {c:i for i,c in enumerate(unique_classes)}
idx_to_class = {v:k for k,v in class_to_idx.items()}

print("\n[2] Exact duplicate (md5) scan...")
hash_map = defaultdict(list)
for p in all_images:
    h = file_md5(p)
    hash_map[h].append(p)
dup_groups = {h:paths for h,paths in hash_map.items() if len(paths) > 1}
n_dup_files = sum(len(v) for v in dup_groups.values())
print("Duplicate groups:", len(dup_groups))
print("Duplicate files (total occurrences in dup groups):", n_dup_files)

print("\n[3] Near-duplicate (pHash) clustering...")
clusters = build_near_duplicate_clusters(all_images, threshold=config.phash_distance_threshold)
nd_clusters = {k:v for k,v in clusters.items() if len(v) > 1}
print("Near-duplicate clusters (size>1):", len(nd_clusters))

print("\n[4] Building group ids for ALL images...")
all_groups = {}
for p in all_images:
    all_groups[p] = p

for cluster_id, members in nd_clusters.items():
    for p in members:
        all_groups[p] = cluster_id

groups_all = [all_groups[p] for p in all_images]

print("\n[5] Group-wise train/test split...")
gss = GroupShuffleSplit(test_size=0.2, n_splits=1, random_state=SEED)
train_idx, test_idx = next(gss.split(all_images, labels_all, groups_all))

train_imgs = [all_images[i] for i in train_idx]
test_imgs  = [all_images[i] for i in test_idx]
train_labels = [labels_all[i] for i in train_idx]
test_labels  = [labels_all[i] for i in test_idx]

print("Train:", len(train_imgs), "Test:", len(test_imgs))

print("Exact overlap train-test:", len(set(train_imgs) & set(test_imgs)))
train_groups = set([groups_all[i] for i in train_idx])
test_groups  = set([groups_all[i] for i in test_idx])
print("Group overlap train-test:", len(train_groups & test_groups))

print("\n[6] Group-wise train/val split...")
train_groups_list = [all_groups[p] for p in train_imgs]
gss2 = GroupShuffleSplit(test_size=0.1, n_splits=1, random_state=SEED)
tr_idx, val_idx = next(gss2.split(train_imgs, train_labels, train_groups_list))

tr_imgs  = [train_imgs[i] for i in tr_idx]
val_imgs = [train_imgs[i] for i in val_idx]

print("Train(core):", len(tr_imgs), "Val:", len(val_imgs))

train_ds = FileListDataset(tr_imgs, class_to_idx, transform=train_transform)
val_ds   = FileListDataset(val_imgs, class_to_idx, transform=eval_transform)
test_ds  = FileListDataset(test_imgs, class_to_idx, transform=eval_transform)

train_loader = DataLoader(train_ds, batch_size=config.batch_size, shuffle=True, num_workers=2, pin_memory=True)
val_loader   = DataLoader(val_ds, batch_size=config.batch_size, shuffle=False, num_workers=2, pin_memory=True)
test_loader  = DataLoader(test_ds, batch_size=config.batch_size, shuffle=False, num_workers=2, pin_memory=True)

# ----------------------------
# Training helpers (ORIGINAL - UNCHANGED)
# ----------------------------
def train_one_epoch(model, loader, optimizer, criterion):
    model.train()
    total_loss, correct, total = 0.0, 0, 0
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        optimizer.zero_grad()
        logits = model(x)
        loss = criterion(logits, y)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * x.size(0)
        pred = logits.argmax(dim=1)
        correct += (pred == y).sum().item()
        total += y.size(0)
    return total_loss / total, correct / total

@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()
    total_loss, correct, total = 0.0, 0, 0
    for x, y in loader:
        x, y = x.to(device), y.to(device)
        logits = model(x)
        loss = criterion(logits, y)
        total_loss += loss.item() * x.size(0)
        pred = logits.argmax(dim=1)
        correct += (pred == y).sum().item()
        total += y.size(0)
    return total_loss / total, correct / total

# ----------------------------
# NEW: Visualization Functions
# ----------------------------
def visualize_attention_maps(model, loader, class_names, num_samples=8):
    """Visualize cross-modal attention maps"""
    model.eval()
    
    for x, y in loader:
        x = x.to(device)
        y = y.numpy()
        break
    
    x = x[:num_samples]
    y = y[:num_samples]
    
    with torch.no_grad():
        _, cnn_attn, vit_attn, fusion_weights = model(x, return_attention=True)
    
    fusion_weights = fusion_weights.cpu().numpy()
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    
    fig, axes = plt.subplots(num_samples, 3, figsize=(15, 5*num_samples))
    if num_samples == 1:
        axes = axes.reshape(1, -1)
    
    for idx in range(num_samples):
        img = x[idx].cpu().numpy().transpose(1, 2, 0)
        img = img * std + mean
        img = np.clip(img, 0, 1)
        
        axes[idx, 0].imshow(img)
        axes[idx, 0].set_title(f'Original\nTrue: {class_names[y[idx]]}', fontsize=10)
        axes[idx, 0].axis('off')
        
        cnn_weight = fusion_weights[idx, 0]
        vit_weight = fusion_weights[idx, 1]
        
        axes[idx, 1].barh(['CNN', 'ViT'], [cnn_weight, vit_weight], color=['#3498db', '#e74c3c'])
        axes[idx, 1].set_xlim([0, 1])
        axes[idx, 1].set_title('Fusion Weights', fontsize=10)
        axes[idx, 1].set_xlabel('Weight')
        
        attn_data = np.array([[cnn_weight, vit_weight]])
        im = axes[idx, 2].imshow(attn_data, cmap='viridis', aspect='auto')
        axes[idx, 2].set_xticks([0, 1])
        axes[idx, 2].set_xticklabels(['CNN→ViT', 'ViT→CNN'])
        axes[idx, 2].set_yticks([])
        axes[idx, 2].set_title('Cross-Modal Attention', fontsize=10)
        plt.colorbar(im, ax=axes[idx, 2], fraction=0.046)
    
    plt.tight_layout()
    plt.savefig('attention_maps.png', dpi=300, bbox_inches='tight')
    plt.show()
    print("✓ Attention maps saved")

@torch.no_grad()
def comprehensive_eval(model, loader, class_names):
    model.eval()
    all_preds, all_true = [], []
    all_probs = []
    all_unc = []

    for x, y in loader:
        x = x.to(device)
        probs, unc = model.predict_with_uncertainty(x, mc_samples=config.mc_dropout_samples)
        pred = probs.argmax(dim=1).cpu().numpy()

        all_preds.extend(pred)
        all_true.extend(y.numpy())
        all_probs.append(probs.cpu().numpy())
        all_unc.extend(unc.cpu().numpy())

    all_preds = np.array(all_preds)
    all_true  = np.array(all_true)
    all_probs = np.vstack(all_probs)
    all_unc   = np.array(all_unc)

    acc = accuracy_score(all_true, all_preds)
    prec, rec, f1, _ = precision_recall_fscore_support(all_true, all_preds, average="weighted")

    print("\n" + "="*60)
    print("TEST EVALUATION (Leakage-mitigated split)")
    print("="*60)
    print(f"Accuracy:  {acc*100:.2f}%")
    print(f"Precision: {prec:.4f}")
    print(f"Recall:    {rec:.4f}")
    print(f"F1-score:  {f1:.4f}")
    
    print("\n" + "-"*60)
    print("UNCERTAINTY STATISTICS")
    print("-"*60)
    print(f"Mean uncertainty: {all_unc.mean():.6f}")
    print(f"Std uncertainty:  {all_unc.std():.6f}")
    print(f"Min uncertainty:  {all_unc.min():.6f}")
    print(f"Max uncertainty:  {all_unc.max():.6f}")
    
    correct_mask = (all_preds == all_true)
    if correct_mask.sum() > 0:
        print(f"\nCorrect predictions - Mean uncertainty: {all_unc[correct_mask].mean():.6f}")
    if (~correct_mask).sum() > 0:
        print(f"Wrong predictions - Mean uncertainty:   {all_unc[~correct_mask].mean():.6f}")

    print("\n" + "-"*60)
    print("ROC-AUC SCORES")
    print("-"*60)
    
    y_true_onehot = np.eye(len(class_names))[all_true]
    auc_macro = roc_auc_score(y_true_onehot, all_probs, average='macro', multi_class='ovr')
    print(f"Macro AUC: {auc_macro:.4f}")
    auc_micro = roc_auc_score(y_true_onehot, all_probs, average='micro', multi_class='ovr')
    print(f"Micro AUC: {auc_micro:.4f}")
    
    print("\nPer-class AUC:")
    for i, class_name in enumerate(class_names):
        try:
            auc_class = roc_auc_score((all_true == i).astype(int), all_probs[:, i])
            print(f"  {class_name}: {auc_class:.4f}")
        except:
            print(f"  {class_name}: N/A")

    print("\n" + "-"*60)
    print("PER-CLASS CLASSIFICATION REPORT")
    print("-"*60)
    print(classification_report(all_true, all_preds, target_names=class_names, digits=4))

    # Confusion Matrix
    cm = confusion_matrix(all_true, all_preds)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=class_names, yticklabels=class_names,
                cbar_kws={'label': 'Count'})
    plt.title("Confusion Matrix", fontsize=16, fontweight='bold')
    plt.ylabel("True Label", fontsize=12)
    plt.xlabel("Predicted Label", fontsize=12)
    plt.tight_layout()
    plt.savefig('confusion_matrix.png', dpi=300, bbox_inches='tight')
    plt.show()

    # ROC Curves
    plt.figure(figsize=(12, 10))
    for i, class_name in enumerate(class_names):
        fpr, tpr, _ = roc_curve((all_true == i).astype(int), all_probs[:, i])
        roc_auc = auc(fpr, tpr)
        plt.plot(fpr, tpr, lw=2, label=f'{class_name} (AUC = {roc_auc:.3f})')
    plt.plot([0, 1], [0, 1], 'k--', lw=2, label='Random Classifier')
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel('False Positive Rate', fontsize=12)
    plt.ylabel('True Positive Rate', fontsize=12)
    plt.title('ROC Curves - Multi-class Classification', fontsize=16, fontweight='bold')
    plt.legend(loc="lower right", fontsize=10)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig('roc_curves.png', dpi=300, bbox_inches='tight')
    plt.show()

    # Uncertainty Distribution
    plt.figure(figsize=(10, 6))
    if correct_mask.sum() > 0:
        plt.hist(all_unc[correct_mask], bins=40, alpha=0.6, label='Correct', 
                 color='green', edgecolor='black')
    if (~correct_mask).sum() > 0:
        plt.hist(all_unc[~correct_mask], bins=40, alpha=0.6, label='Wrong', 
                 color='red', edgecolor='black')
    plt.xlabel('Uncertainty (Variance)', fontsize=12)
    plt.ylabel('Frequency', fontsize=12)
    plt.title('Uncertainty Distribution', fontsize=14, fontweight='bold')
    plt.legend(fontsize=11)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig('uncertainty_distribution.png', dpi=300, bbox_inches='tight')
    plt.show()

    return {
        "accuracy": acc,
        "precision": prec,
        "recall": rec,
        "f1": f1,
        "auc_macro": auc_macro,
        "auc_micro": auc_micro,
        "y_true": all_true,
        "y_pred": all_preds,
        "y_probs": all_probs,
        "uncertainty": all_unc,
    }

def bootstrap_ci(y_true, y_pred, metric_fn, n=1000):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    stats = []
    for _ in range(n):
        idx = np.random.choice(len(y_true), len(y_true), replace=True)
        stats.append(metric_fn(y_true[idx], y_pred[idx]))
    return np.percentile(stats, [2.5, 97.5])

def plot_bootstrap_ci(y_true, y_pred, n_bootstrap=1000):
    """Visualize bootstrap confidence intervals"""
    acc_samples = []
    for _ in range(n_bootstrap):
        idx = np.random.choice(len(y_true), len(y_true), replace=True)
        acc_samples.append(accuracy_score(y_true[idx], y_pred[idx]))
    
    acc_samples = np.array(acc_samples)
    acc_mean = acc_samples.mean()
    acc_ci = np.percentile(acc_samples, [2.5, 97.5])
    
    plt.figure(figsize=(12, 5))
    
    plt.subplot(1, 2, 1)
    plt.hist(acc_samples, bins=50, color='steelblue', edgecolor='black', alpha=0.7)
    plt.axvline(acc_mean, color='red', linestyle='--', linewidth=2, label=f'Mean: {acc_mean:.4f}')
    plt.axvline(acc_ci[0], color='orange', linestyle='--', linewidth=2, label=f'95% CI: [{acc_ci[0]:.4f}, {acc_ci[1]:.4f}]')
    plt.axvline(acc_ci[1], color='orange', linestyle='--', linewidth=2)
    plt.xlabel('Accuracy', fontsize=12)
    plt.ylabel('Frequency', fontsize=12)
    plt.title('Bootstrap Distribution', fontsize=14, fontweight='bold')
    plt.legend(fontsize=10)
    plt.grid(alpha=0.3)
    
    plt.subplot(1, 2, 2)
    y_pos = [0]
    plt.barh(y_pos, [acc_mean], xerr=[[acc_mean - acc_ci[0]], [acc_ci[1] - acc_mean]],
             color='steelblue', alpha=0.7, capsize=10)
    plt.yticks(y_pos, ['Accuracy'])
    plt.xlabel('Score', fontsize=12)
    plt.title('95% Confidence Interval', fontsize=14, fontweight='bold')
    plt.xlim([0, 1])
    plt.grid(axis='x', alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('bootstrap_confidence_intervals.png', dpi=300, bbox_inches='tight')
    plt.show()
    print(f"\nAccuracy 95% CI: [{acc_ci[0]*100:.2f}%, {acc_ci[1]*100:.2f}%]")

def plot_training_curves(history):
    """Plot training and validation curves"""
    epochs_range = range(1, len(history['train_loss']) + 1)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
    
    ax1.plot(epochs_range, history['train_loss'], 'b-', linewidth=2, label='Training Loss')
    ax1.plot(epochs_range, history['val_loss'], 'r-', linewidth=2, label='Validation Loss')
    ax1.set_xlabel('Epoch', fontsize=12)
    ax1.set_ylabel('Loss', fontsize=12)
    ax1.set_title('Training and Validation Loss', fontsize=14, fontweight='bold')
    ax1.legend(fontsize=11)
    ax1.grid(alpha=0.3)
    
    ax2.plot(epochs_range, history['train_acc'], 'b-', linewidth=2, label='Training Accuracy')
    ax2.plot(epochs_range, history['val_acc'], 'r-', linewidth=2, label='Validation Accuracy')
    ax2.set_xlabel('Epoch', fontsize=12)
    ax2.set_ylabel('Accuracy', fontsize=12)
    ax2.set_title('Training and Validation Accuracy', fontsize=14, fontweight='bold')
    ax2.legend(fontsize=11)
    ax2.grid(alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('training_curves.png', dpi=300, bbox_inches='tight')
    plt.show()

# ----------------------------
# Main training loop (ORIGINAL - UNCHANGED)
# ----------------------------
def main():
    print("\n" + "="*60)
    print("TRAINING: HybridEnsemble (Leakage-mitigated)")
    print("="*60)

    class_names = [idx_to_class[i] for i in range(len(idx_to_class))]

    model = HybridEnsemble(
        num_classes=config.num_classes,
        pretrained=True,
        fusion_dim=config.fusion_dim,
        dropout_rate=config.dropout_rate
    ).to(device)

    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"Total params: {total_params:,} | Trainable: {trainable_params:,}")

    criterion = FocalLossWithSmoothing(alpha=1.0, gamma=2.0, smoothing=0.1)
    optimizer = torch.optim.AdamW(model.parameters(), lr=config.lr, weight_decay=config.weight_decay)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=config.epochs)

    best_val_acc = 0.0
    
    history = {
        'train_loss': [],
        'train_acc': [],
        'val_loss': [],
        'val_acc': []
    }

    for epoch in range(1, config.epochs + 1):
        tr_loss, tr_acc = train_one_epoch(model, train_loader, optimizer, criterion)
        va_loss, va_acc = evaluate(model, val_loader, criterion)
        scheduler.step()
        
        history['train_loss'].append(tr_loss)
        history['train_acc'].append(tr_acc)
        history['val_loss'].append(va_loss)
        history['val_acc'].append(va_acc)

        print(f"Epoch {epoch:02d}/{config.epochs} | "
              f"Train loss {tr_loss:.4f} acc {tr_acc:.4f} | "
              f"Val loss {va_loss:.4f} acc {va_acc:.4f}")

        if va_acc > best_val_acc:
            best_val_acc = va_acc
            torch.save(model.state_dict(), "best_hybrid_ensemble_leakage_free.pth")
            print(f"  -> saved best (val acc {best_val_acc:.4f})")

    print("\nPlotting training curves...")
    plot_training_curves(history)

    print("\nLoading best model for testing...")
    model.load_state_dict(torch.load("best_hybrid_ensemble_leakage_free.pth", map_location=device))
    
    print("\nVisualizing attention maps...")
    visualize_attention_maps(model, test_loader, class_names, num_samples=8)
    
    results = comprehensive_eval(model, test_loader, class_names)

    print("\nComputing bootstrap confidence intervals...")
    plot_bootstrap_ci(results["y_true"], results["y_pred"], n_bootstrap=1000)

    return results

if __name__ == "__main__":
    results = main()
    print("\n" + "="*60)
    print("EXPERIMENT FINISHED SUCCESSFULLY ✅")
    print("="*60)
    print("\nGenerated visualizations:")
    print("  1. confusion_matrix.png")
    print("  2. roc_curves.png")
    print("  3. training_curves.png")
    print("  4. attention_maps.png")
    print("  5. uncertainty_distribution.png")
    print("  6. bootstrap_confidence_intervals.png")
    print("="*60)
```
